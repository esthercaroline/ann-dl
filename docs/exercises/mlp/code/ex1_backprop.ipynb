# Exercise 1 — One Step of Backpropagation (numerical check)

The derivation is done by hand in the report. This notebook only re-computes
every intermediate value with plain NumPy so the hand-written numbers can be
checked. No library model or autodiff is used.

```python
import numpy as np

np.set_printoptions(precision=6, suppress=True)

x = np.array([0.5, -0.2])
y = 1.0
W1 = np.array([[0.3, -0.1], [0.2, 0.4]])
b1 = np.array([0.1, -0.2])
W2 = np.array([0.5, -0.3])
b2 = 0.2
eta = 0.3
```

## A — Forward pass

```python
z1 = W1 @ x + b1
h1 = np.tanh(z1)
u2 = W2 @ h1 + b2
y_hat = np.tanh(u2)
L = (y - y_hat) ** 2
print("z1    =", z1)
print("h1    =", h1)
print(f"u2    = {u2:.6f}")
print(f"y_hat = {y_hat:.6f}")
print(f"L     = {L:.6f}")
```

## B — Backward pass

```python
dL_dyhat = -2 * (y - y_hat)
dL_du2 = dL_dyhat * (1 - y_hat ** 2)
dL_dW2 = dL_du2 * h1
dL_db2 = dL_du2
dL_dh1 = dL_du2 * W2
dL_dz1 = dL_dh1 * (1 - h1 ** 2)
dL_dW1 = np.outer(dL_dz1, x)
dL_db1 = dL_dz1
print(f"dL/dyhat = {dL_dyhat:.6f}")
print(f"dL/du2   = {dL_du2:.6f}")
print("dL/dW2   =", dL_dW2)
print(f"dL/db2   = {dL_db2:.6f}")
print("dL/dh1   =", dL_dh1)
print("dL/dz1   =", dL_dz1)
print("dL/dW1   =\n", dL_dW1)
print("dL/db1   =", dL_db1)
```

## C — Parameter update and a second forward pass

```python
W2_new = W2 - eta * dL_dW2
b2_new = b2 - eta * dL_db2
W1_new = W1 - eta * dL_dW1
b1_new = b1 - eta * dL_db1
print("W2 ->", W2_new)
print(f"b2 -> {b2_new:.6f}")
print("W1 ->\n", W1_new)
print("b1 ->", b1_new)

y_hat_2 = np.tanh(W2_new @ np.tanh(W1_new @ x + b1_new) + b2_new)
print(f"\nsecond pass: y_hat = {y_hat_2:.6f}, L = {(y - y_hat_2) ** 2:.6f}")
```
